```python
# importing numpy and using it as np.method_name()
import numpy as np
```

```python
# Creating float numpy arrays from a list
python_list = [1,2,3,4,5,6]
```

```python
print(type(python_list))
```

```python
# converting the above python list into an array(numpy)
numpy_array_from_list = np.array(python_list, dtype=float)
print(type(numpy_array_from_list))
print(f"Before converting into numpy array: {python_list} --> gives list")
print(f"After converting inot numpy array: {numpy_array_from_list} --> gives numpy array")
```

```python
# Creating numpy array from a tuple
python_tuple = (1, 2, 3, 4, 5)
print(type(python_tuple))
print(f"Python tuple: {python_tuple}")
```

```python
# Coverting the above python tuple into an array(numpy)
numpy_array_from_tuple = np.array(python_tuple)
print(type(numpy_array_from_tuple))
print(f"After converting into numpy array: {numpy_array_from_tuple}")
```
