```python
def count_lines_and_words(filename):
    with open(filename, "r") as file:
        no_of_lines = 0
        no_of_words = 0

        for line in file:
            no_of_lines += 1
            no_of_words += len(line.split())

    return no_of_lines, no_of_words

obama_speech = count_lines_and_words("./data/obama_speech.txt")
michelle_obama_speech = count_lines_and_words("./data/melina_trump_speech.txt")
donald_speech = count_lines_and_words("./data/donald_speech.txt")
melina_trump_speech = count_lines_and_words("./data/melina_trump_speech.txt")

print(obama_speech)
print(michelle_obama_speech)
print(donald_speech)
print(melina_trump_speech)
```

```python
import json

def most_spoken_languages(filename, n):
    with open(filename, encoding="utf-8") as file: # this line will automatically close the file after all operations are done the default mode after opening the file is read mode.
        countries_data = json.load(file)
        language_counts = {}
        for country in countries_data:
            for language in country["languages"]:
                if language in language_counts:
                    language_counts[language] += 1
                else:
                    language_counts[language] = 1
    
        sorted_languages = sorted(language_counts.items(), key = lambda item: item[1], reverse=True)
        final_result = [(count, language) for language, count in sorted_languages]
        return final_result[:n]

print(most_spoken_languages("./data/countries_data.json", 3))
print(most_spoken_languages("./data/countries_data.json", 10))
```

```python
def most_populated_country(filename, n):
    with open(filename, encoding="utf-8") as file:
        countries_data = json.load(file)

        sorted_countries = sorted(countries_data, key = lambda country: country["population"], reverse=True)

        result = [
            {
                "country": country["name"],
                "population": country["population"]
            }
            for country in sorted_countries
        ]

        return result[:n]

most_populated_country("./data/countries_data.json", 3)
```

```python
import re

def extract_email_addresses(filename):
    with open(filename) as file:
        content = file.read()
        pattern = r"[\w\.-]+@[\w\.-]+\.\w+"
        email_addresses = re.findall(pattern, content)
        return email_addresses

extract_email_addresses("./data/email_exchange_big.txt")
```
