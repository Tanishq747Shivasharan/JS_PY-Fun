```python
def count_lines_and_words(filename):
    with open(filename, "r") as file:
        no_of_lines = 0
        no_of_words = 0

        for line in file:
            no_of_lines += 1
            no_of_words += len(line.split())

    return no_of_lines, no_of_words

obama_speech = count_lines_and_words("./data/obama_speech.txt")
michelle_obama_speech = count_lines_and_words("./data/melina_trump_speech.txt")
donald_speech = count_lines_and_words("./data/donald_speech.txt")
melina_trump_speech = count_lines_and_words("./data/melina_trump_speech.txt")

print(obama_speech)
print(michelle_obama_speech)
print(donald_speech)
print(melina_trump_speech)
```
