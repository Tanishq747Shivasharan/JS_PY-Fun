```python
def count_lines_and_words(filename):
    with open(filename, "r") as file:
        no_of_lines = 0
        no_of_words = 0

        for line in file:
            no_of_lines += 1
            no_of_words += len(line.split())

    return no_of_lines, no_of_words

obama_speech = count_lines_and_words("./data/obama_speech.txt")
michelle_obama_speech = count_lines_and_words("./data/melina_trump_speech.txt")
donald_speech = count_lines_and_words("./data/donald_speech.txt")
melina_trump_speech = count_lines_and_words("./data/melina_trump_speech.txt")

print(obama_speech)
print(michelle_obama_speech)
print(donald_speech)
print(melina_trump_speech)
```

```python
import json

def most_spoken_languages(filename, n):
    with open(filename, encoding="utf-8") as file: # this line will automatically close the file after all operations are done the default mode after opening the file is read mode.
        countries_data = json.load(file)
        language_counts = {}
        for country in countries_data:
            for language in country["languages"]:
                if language in language_counts:
                    language_counts[language] += 1
                else:
                    language_counts[language] = 1
    
        sorted_languages = sorted(language_counts.items(), key = lambda item: item[1], reverse=True)
        final_result = [(count, language) for language, count in sorted_languages]
        return final_result[:n]

print(most_spoken_languages("./data/countries_data.json", 3))
print(most_spoken_languages("./data/countries_data.json", 10))
```

```python
def most_populated_country(filename, n):
    with open(filename, encoding="utf-8") as file:
        countries_data = json.load(file)

        sorted_countries = sorted(countries_data, key = lambda country: country["population"], reverse=True)

        result = [
            {
                "country": country["name"],
                "population": country["population"]
            }
            for country in sorted_countries
        ]

        return result[:n]

most_populated_country("./data/countries_data.json", 3)
```

```python
import re

def extract_email_addresses(filename):
    with open(filename) as file:
        content = file.read()
        pattern = r"[\w\.-]+@[\w\.-]+\.\w+"
        email_addresses = re.findall(pattern, content)
        return email_addresses

extract_email_addresses("./data/email_exchange_big.txt")
```

```python
import re

def find_most_common_words(filename, n):
    with open(filename) as file:
        text = file.read().lower()
        pattern = r"\b\w+\b"
        words = re.findall(pattern, text)
        word_counts = {}
        for word in words:
            if word in word_counts:
                word_counts[word] += 1
            else:
                word_counts[word] = 1

        sorted_words = sorted(word_counts.items(), key=lambda item: item[1], reverse=True)
        final_result = [(count, word) for word, count in sorted_words]
        return final_result[:n]

sample_file = find_most_common_words("./data/sample.txt", 7)
obama_speech = find_most_common_words("./data/obama_speech.txt", 10)
michelle_obama_speech = find_most_common_words("./data/michelle_obama_speech.txt", 10)
donald_speech = find_most_common_words("./data/donald_speech.txt", 10)
melina_trump_speech = find_most_common_words("./data/melina_trump_speech.txt", 10)
top_10 = find_most_common_words("./data/romeo_and_juliet.txt", 10)

print(sample_file)
print(obama_speech)
print(michelle_obama_speech)
print(donald_speech)
print(melina_trump_speech)
print(top_10)
```

```python
import re
from data.stop_words import stop_words

def clean_text(filename):
    with open(filename) as file:
        text = file.read().lower()
        pattern = r"[^\s\w]"
        cleaned_txt = re.sub(pattern, '', text)
        return cleaned_txt

cleaned_text = clean_text("./data/sample.txt")

def remove_support_words(cleaned_text):
    words = cleaned_text.split()
    result = []
    for word in words:
        if word not in stop_words:
            result.append(word)

    return result

def check_text_similarity(speech1, speech2):
    common = 0

    for word in speech1:
        if word in speech2:
            common += 1

    similarity = (common / len(speech1)) * 100

    return similarity

speech1 = remove_support_words(clean_text("./data/michelle_obama_speech.txt"))
speech2 = remove_support_words(clean_text("./data/melina_trump_speech.txt"))

print(check_text_similarity(speech1, speech2))
```
