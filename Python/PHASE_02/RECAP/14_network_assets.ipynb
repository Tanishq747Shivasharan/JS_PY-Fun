OOP: Classes, Instances, and Inheritance

```python
class Statistics:
    def __init__(self, data):
        self.data = data

    def count(self):
        return len(self.data)

    def total(self):
        return sum(self.data)

    def min(self):
        return min(self.data)

    def max(self):
        return max(self.data)

    def range(self):
        return self.max() - self.min()

    def mean(self):
        return round(self.total() / self.count())

    def median(self):
        sorted_data = sorted(self.data)
        n = self.count()
        middle = n // 2

        if n % 2 == 1:
            return sorted_data[middle]
        else:
            return (sorted_data[middle - 1] + sorted_data[middle]) / 2

    def mode(self):
        frequency = {}

        for value in self.data:
            if value in frequency:
                frequency[value] += 1
            else:
                frequency[value] = 1

        mode_value = max(frequency, key=frequency.get)

        return mode_value, frequency[mode_value]

    def var(self):
        mean = self.total() / self.count()

        squared_differences = []

        for value in self.data:
            squared_differences.append((value - mean) ** 2)

        return sum(squared_differences) / (self.count() - 1)

    def std(self):
        return self.var() ** 0.5

    def freq_dist(self):
        frequency = {}

        for value in self.data:
            frequency[value] = frequency.get(value, 0) + 1

        result = []

        for value, count in frequency.items():
            percentage = (count / self.count()) * 100
            result.append((percentage, value))

        result.sort(reverse=True)

        return result

    def describe(self):
        return (
            f"Count: {self.count()}\n"
            f"Sum: {self.total()}\n"
            f"Min: {self.min()}\n"
            f"Max: {self.max()}\n"
            f"Range: {self.range()}\n"
            f"Mean: {self.mean()}\n"
            f"Median: {self.median()}\n"
            f"Mode: {self.mode()}\n"
            f"Variance: {round(self.var(), 1)}\n"
            f"Standard Deviation: {round(self.std(), 1)}\n"
            f"Frequency Distribution: {self.freq_dist()}"
        )

ages = [31, 26, 34, 37, 27, 26, 32, 32, 26, 27, 27, 24, 32, 33, 27, 25, 26, 38, 37, 31, 34, 24, 33, 29, 26]

data = Statistics(ages)

print(data.describe())
```

```python
class PersonAccount:
    def __init__(self, firstname, lastname):
        self.firstname = firstname
        self.lastname = lastname
        self.incomes = {}
        self.expenses = {}

    def add_income(self, description, amount):
        self.incomes[description] = amount
    
    def add_expense(self, description, amount):
        self.expenses[description] = amount
    
    def total_income(self):
        return sum(self.incomes.values())

    def total_expense(self):
        return sum(self.expenses.values())

    def account_balance(self):
        return self.total_income() - self.total_expense()
    
    def account_info(self):
        return f"Account Information\nFirstname: {self.firstname}\nLastname: {self.lastname}\nTotal income: {self.total_income()}\nTotal expenses: {self.total_expense()}\nAccount balance: {self.account_balance()}"
        

Minato = PersonAccount("Minato", "Namikaze")

Minato.add_income("Salary", 50000)
Minato.add_income("Freelancing", 10000)

Minato.add_expense("Rent", 15000)
Minato.add_expense("Food", 5000)

print(Minato.account_info())
```

```python
import requests
from bs4 import BeautifulSoup
import json

url = 'http://www.bu.edu/president/boston-university-facts-stats/'

# Get webpage
response = requests.get(url)

# Check response
print("Status Code:", response.status_code)

# Parse HTML
content = response.content
soup = BeautifulSoup(content, 'html.parser')

data = {}

sections = soup.find_all('section', class_='stat-section')

for section in sections:
    title = section.find('h4').get_text(strip=True)

    labels = section.find_all('span', class_='stat-label')
    figures = section.find_all('span', class_='stat-figure')

    data[title] = {}

    for label, figure in zip(labels, figures):
        data[title][label.get_text(strip=True)] = figure.get_text(strip=True)

campus_heading = soup.find(
    'h4',
    class_='stat-group-title',
    string=lambda text: text and text.strip() == 'Campus'
)

campus_container = campus_heading.find_next(
    'div',
    class_='bu-stat-list'
)

campus_titles = campus_container.find_all(
    'h3',
    class_='bu-stat-title'
)

campus_values = campus_container.find_all(
    'span',
    class_='bu-stat-value-field'
)

data['Campus'] = {}

for title, value in zip(campus_titles, campus_values):
    data['Campus'][title.get_text(strip=True)] = value.get_text(strip=True)

with open('bu_facts_stats.json', 'w', encoding='utf-8') as file:
    json.dump(data, file, indent=4)

print("Data saved successfully!")
```

```python
import requests
import json
from bs4 import BeautifulSoup

url = 'https://archive.ics.uci.edu/datasets'

response = requests.get(url)

print(response.status_code)

soup = BeautifulSoup(response.content, 'html.parser')
t = soup.find('table')
rows = t.find_all('tr')

data = []

for row in rows:
    cells = row.find_all('td')
    data.append([cell.get_text(strip=True) for cell in cells])

with open('uci_datasets.json', 'w') as file:
    json.dump(data, file, indent=4)

print("Data saved successfully!")
```

```python
import requests
from bs4 import BeautifulSoup
import json

url = 'https://en.wikipedia.org/wiki/List_of_presidents_of_the_United_States'

headers = {
    'User-Agent': 'Mozilla/5.0'
}

response = requests.get(url, headers=headers)

soup = BeautifulSoup(response.content, 'html.parser')

tables = soup.find_all('table')

table = tables[0]

rows = table.find_all('tr')

data = []

for row in rows:
    cells = row.find_all(['th', 'td'])
    data.append([cell.get_text(" ", strip=True) for cell in cells])

with open('presidents.json', 'w', encoding='utf-8') as file:
    json.dump(data, file, indent=4)
```
